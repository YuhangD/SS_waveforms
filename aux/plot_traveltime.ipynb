```python
from obspy.taup import plot_travel_times
import matplotlib.pyplot as plt
```

```python
fig, ax  = plt.subplots()
ax = plot_travel_times(model = "prem",source_depth = 75, ax = ax, fig = fig, 
                       phase_list = ['s','S','SS', 'S^410S','S^660S','sS^410S'])
                    #    phase_list = ['p','P','pP','s','S','sS','pS','sP', 'PcP', 'ScP'])
```

```python
import numpy as np
S_data = np.load('S_traveltime.npz')
fig, ax  = plt.subplots()
ax = plot_travel_times(source_depth = 75, ax = ax, fig = fig, 
                       phase_list = ['s','S','SS'])
id = np.argwhere(S_data['depths']==75)
id = int(id.squeeze())

ax.plot(S_data['gcarcs'][:],S_data['timetable'][:,id]/60, lw = 2, color = 'black', label = 'S_arr')

plt.show()
```
