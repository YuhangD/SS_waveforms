```python
import os
import numpy as np
import pandas as pd
```

```python
model = 'IASP91'
dataroot = './'

ntrace = np.zeros([2,1146])

for ibin in range(1,1147):
    
    df = pd.read_excel("/Users/tdgb464/OneDrive - University of Leeds/MTZ_SS/grid_6/Migrations/"+model+"/Info/Bin"+str(ibin)+"_info.xlsx")
    new_df = pd.DataFrame(columns = df.columns)
    for i, row in df.iterrows():
        year = row.event.split('_')[1]
        event = row.event.split('/')[0]
        netsta = row.event.split('/')[1].strip('_')
        filename = dataroot+"/"+year+"/"+event+"/"+netsta+".BHT.sac"
        if os.path.isfile(filename):
            new_df = pd.concat([new_df, pd.DataFrame(row).T], ignore_index = True)
    
    new_df.to_excel("./Info/Bin"+str(ibin)+"_info.xlsx", index = False)
    ntrace[0][ibin - 1] = len(new_df)
    ntrace[1][ibin - 1] = len(df)
    print(f"{ibin}: {len(new_df)}/{len(df)}")
```
